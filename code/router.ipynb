```python
from pathlib import Path
DATASET_DIR = Path.cwd().parent / "dataset" 
DATASET_DIR
```

```python
import pandas as pd

business_acc_data = pd.read_csv(str(DATASET_DIR) + "/business_accounts.csv")
daily_notification_data = pd.read_csv(str(DATASET_DIR) + "/daily_notification_summary.csv")
group_members_data = pd.read_csv(str(DATASET_DIR) + "/group_members.csv")
groups_data = pd.read_csv(str(DATASET_DIR) + "/groups.csv")
images_data = pd.read_csv(str(DATASET_DIR) + "/images.csv")
message_events_data = pd.read_csv(str(DATASET_DIR) + "/message_events.csv")
message_history_data = pd.read_csv(str(DATASET_DIR) + "/message_history.csv")
user_business_history_data = pd.read_csv(str(DATASET_DIR) + "/user_business_history.csv")
users_data = pd.read_csv(str(DATASET_DIR) + "/users.csv")
voice_notes_data = pd.read_csv(str(DATASET_DIR) + "/voice_notes.csv")
messages_data = pd.read_csv(str(DATASET_DIR) + "/messages.csv")
```

```python
messages_df = messages_data.copy(deep=True)
messages_df.rename(columns={"created_at":"message_created_at","media_id":"message_media_id","media_type":"message_media_type","forwarded_count":"message_forwarded_count", "conversation_type":"message_conversation_type"}, inplace=True)
users_df = users_data.copy(deep=True)
users_df.rename(columns={"notifications_dismissed_30d":"total_notifications_dismissed_30d","messages_reported_30d":"total_messages_reported_30d","do_not_disturb_window":"users_do_not_disturb_window", "messages_opened_30d":"users_total_messages_opened_30d","messages_replied_30d":"users_total_messages_replied_30d","notifications_dismissed_30d":"users_total_notifications_dismissed_30d", "messages_reported_30d":"users_total_messages_reported_30d"}, inplace=True)
messages_df = messages_df.merge(users_df, how="left", on="user_id")
messages_df.head()
```

```python
# Notification load over the latest seven days available
notification_summary = daily_notification_data.copy()
notification_summary["date"] = pd.to_datetime(notification_summary["date"])
latest_date = notification_summary["date"].max()
recent_notifications = notification_summary[
    notification_summary["date"] >= latest_date - pd.Timedelta(days=6)
]

notifications_df = (
    recent_notifications.groupby("user_id", as_index=False)
    .agg(
        notifications_sent_7d=("notifications_sent", "sum"),
        notifications_dismissed_7d=("notifications_dismissed", "sum"),
    )
)
```

```python
notifications_df.head()
```

```python
messages_df = messages_df.merge(notifications_df, how="left", on="user_id")
messages_df.columns
```

```python
message_history_df = message_history_data.copy(deep=True)
message_history_df["created_at"] = pd.to_datetime(message_history_df["created_at"])
#message_history_df.head()
message_history_df.rename(columns={"created_at":"message_history_created_at","media_id":"message_history_media_id","media_type":"message_history_media_type","forwarded_count":"message_history_forwarded_count", "conversation_type":"message_history_conversation_type"}, inplace=True)
message_history_df = message_history_df.merge(images_df, how="left", left_on="message_history_media_id", right_on="image_id")
message_history_df = message_history_df.merge(voice_notes_df, how="left", left_on="message_history_media_id", right_on="voice_note_id")
message_history_df.head()
```

```python
message_history_df.head()
```

```python
message_history_df.info()
```

```python
message_events_df = message_events_data.copy(deep=True)
message_events_df.head()
```

```python
message_events_df.rename(columns={"reaction_time_minutes":"message_event_reaction_time_minutes","notification_dismissed":"message_event_notification_dismissed","muted_after_message":"message_event_muted_after_message","message_reported":"message_event_reported"}, inplace=True)
message_events_df.head()
```

```python
message_history_df = message_history_df.merge(message_events_df, how="left", on=["message_id","user_id"])
message_history_df.head()
```

```python
message_history_df.columns
```

```python
cols_to_fill = ["message_event_reaction_time_minutes", "message_event_notification_dismissed", "message_event_muted_after_message", "message_event_reported"]
message_history_df[cols_to_fill] = message_history_df[cols_to_fill].fillna(0).astype(int)
```

```python
message_history_df.info()
```

```python
group_interaction_summary = message_history_df.groupby(["group_id", "user_id"]).agg(
    total_group_messages_sent=("message_id", "count"),
    total_group_messages_opened=("message_opened", "sum"),
    total_group_messages_replied=("message_replied", "sum"),
    total_group_messages_dismissed=("message_event_notification_dismissed", "sum"),
    total_group_messages_muted=("message_event_muted_after_message", "sum"),
    total_group_messages_reported=("message_event_reported", "sum")).reset_index()
group_interaction_summary.head()
```

```python
business_interaction_summary = message_history_df.groupby(["business_id", "user_id"]).agg(
    total_business_messages_sent=("message_id", "count"),
    total_business_messages_opened=("message_opened", "sum"),
    total_business_messages_replied=("message_replied", "sum"),
    total_business_messages_dismissed=("message_event_notification_dismissed", "sum"),
    total_business_messages_muted=("message_event_muted_after_message", "sum"),
    total_business_messages_reported=("message_event_reported", "sum")).reset_index()

business_interaction_summary.head()
```

```python
sender_interaction_summary = message_history_df.groupby(["sender_user_id", "user_id"]).agg(
    total_sender_messages_sent=("message_id", "count"),
    total_sender_messages_opened=("message_opened", "sum"),
    total_sender_messages_replied=("message_replied", "sum"),
    total_sender_messages_dismissed=("message_event_notification_dismissed", "sum"),
    total_sender_messages_muted=("message_event_muted_after_message", "sum"),
    total_sender_messages_reported=("message_event_reported", "sum")).reset_index()

sender_interaction_summary.head()
```

```python
messages_df.info()
```

```python
messages_df = messages_df.merge(group_interaction_summary, how="left", on=["group_id", "user_id"])
messages_df = messages_df.merge(business_interaction_summary, how="left", on=["business_id", "user_id"])
messages_df = messages_df.merge(sender_interaction_summary, how="left", on=["sender_user_id", "user_id"])
messages_df.to_excel(DATASET_DIR / "modified_data" / "final_dataset.xlsx", index=False)
```
